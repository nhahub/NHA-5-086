```python
df_orders = df_orders.drop(columns=["_ingested_at","gender"])
```

```python
date_cols = ["created_at", "shipped_at", "delivered_at", "returned_at"]
```

```python
df = df_orders

invalid_created_shipped = (
    df["created_at"].notna()
    & df["shipped_at"].notna()
    & (df["created_at"] > df["shipped_at"])
)

invalid_shipped_delivered = (
    df["shipped_at"].notna()
    & df["delivered_at"].notna()
    & (df["shipped_at"] > df["delivered_at"])
)

invalid_delivered_returned = (
    df["delivered_at"].notna()
    & df["returned_at"].notna()
    & (df["delivered_at"] > df["returned_at"])
)
```

```python
invalid_rows = (
    invalid_created_shipped | invalid_shipped_delivered | invalid_delivered_returned
)

df_orders.loc[invalid_rows, date_cols]
```

```python
print("Created > Shipped:", invalid_created_shipped.sum())

print("Shipped > Delivered:", invalid_shipped_delivered.sum())

print("Delivered > Returned:", invalid_delivered_returned.sum())

print("Total invalid rows:", invalid_rows.sum())
```

```python
df_orders.to_parquet("../cleaned_data/orders.parquet", index=False)
```
