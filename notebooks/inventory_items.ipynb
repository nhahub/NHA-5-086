```python
product_ids = [
    "10598",
    "11389",
    "11668",
    "11843",
    "12586",
    "13863",
    "13921",
    "15723",
    "15757",
    "1629",
    "16309",
    "16559",
    "16898",
    "21207",
    "21484",
    "23769",
    "24287",
    "24455",
    "25135",
    "25187",
    "27288",
    "27543",
    "27640",
    "755",
    "8600",
    "9482",
]

columns_needed = [
    "product_id",
    "cost",
    "product_category",
    "product_name",
    "product_brand",
    "product_retail_price",
    "product_department",
    "product_distribution_center_id",
]

rename_map = {
    "product_id": "id",
    "product_category": "category",
    "product_name": "name",
    "product_brand": "brand",
    "product_retail_price": "retail_price",
    "product_department": "department",
    "product_distribution_center_id": "distribution_center_id",
}

na_tokens = {"n/a", "na", "none", "null", ""}

string_cols = [
    "id",
    "category",
    "name",
    "brand",
    "department",
    "distribution_center_id",
]

result = (
    df_inventory_items[df_inventory_items["product_id"].astype(str).isin(product_ids)][columns_needed]
    .rename(columns=rename_map)
    .drop_duplicates()
    .reset_index(drop=True)
)

for col in string_cols:
    s = result[col].astype("string")
    mask_na = s.isna() | s.str.strip().str.lower().isin(na_tokens)
    result[col] = s.where(~mask_na, "unknown")
```
