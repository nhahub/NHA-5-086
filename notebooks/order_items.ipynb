```python
from depi_project.etl.bronze.loader import load_table
from depi_project.cleaning.check import *
import pandas as pd
```

```python
df_order_items = load_table("bronze", "order_items")
```

```python
categorical_validity_checks(df_order_items)
get_information(df_order_items)
```

```python
check_missing_values(df_order_items)
check_duplicates(df_order_items)
```

```python
id_cols = ["id", "order_id", "user_id", "product_id", "inventory_item_id"]
df_order_items[id_cols] = df_order_items[id_cols].astype("string")
```

```python
check_data_type(df_order_items)
```

```python
df_order_items = df_order_items.drop(columns=["_ingested_at"])
```

```python
df_order_items
```

```python
df_order_items["sale_price"] = df_order_items["sale_price"].round(2)
df_order_items 
```

```python
date_cols = ["created_at", "shipped_at", "delivered_at", "returned_at"]
```

```python
df = df_order_items

invalid_created_shipped = (
    df["created_at"].notna()
    & df["shipped_at"].notna()
    & (df["created_at"] > df["shipped_at"])
)

invalid_shipped_delivered = (
    df["shipped_at"].notna()
    & df["delivered_at"].notna()
    & (df["shipped_at"] > df["delivered_at"])
)

invalid_delivered_returned = (
    df["delivered_at"].notna()
    & df["returned_at"].notna()
    & (df["delivered_at"] > df["returned_at"])
)
```

```python
invalid_rows = (
    invalid_created_shipped | invalid_shipped_delivered | invalid_delivered_returned
)

df_order_items.loc[invalid_rows, date_cols]
```

```python
print("Created > Shipped:", invalid_created_shipped.sum())

print("Shipped > Delivered:", invalid_shipped_delivered.sum())

print("Delivered > Returned:", invalid_delivered_returned.sum())

print("Total invalid rows:", invalid_rows.sum())
```

```python
df_order_items.drop(columns=["created_at", "shipped_at", "delivered_at", "returned_at"], inplace=True)
```
